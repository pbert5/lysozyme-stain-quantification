```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from functools import partial
from PIL import Image
from IPython.display import display 
from matplotlib import animation


img0_path = '/home/phillip/documents/code/lysozyme/results/rendering_test_run/debug_intermediates/G2FB_-_2_Retake_G2_-_ABX_414a0a/identify_crypt_seeds_new/tissue_caps_input.png'
img1_path = '/home/phillip/documents/code/lysozyme/results/rendering_test_run/debug_intermediates/G2FB_-_2_Retake_G2_-_ABX_414a0a/identify_crypt_seeds_new/crypt_troughs.png'

N = 60  # number of frames
img0 = np.array(Image.open(img0_path))
img1 = np.array(Image.open(img1_path))
```

```python
frames = []
N = 20
for i in range(N + 1):
    alpha = i / N
    blended_img = (1 - alpha) * img0 + alpha * img1
    blended_img = blended_img.astype(np.uint8)
    frames.append(blended_img)



```

```python
from IPython import display

# convert frames to html5 video
fig = plt.figure()
im = plt.imshow(frames[0])
plt.axis('off')

def update_frame(i):
    im.set_array(frames[i])
    return [im]

ani = FuncAnimation(fig, update_frame, frames=len(frames), interval=50, blit=True)

plt.show()
```
