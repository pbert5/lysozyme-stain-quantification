```python
import numpy as np
from cellpose import models, core, io, plot
from pathlib import Path
from tqdm import trange
import matplotlib.pyplot as plt
from natsort import natsorted

io.logger_setup() # run this to get printing of progress

model = models.CellposeModel(gpu=False)
```

```python







flow_threshold = 0.4
cellprob_threshold = 0.0
tile_norm_blocksize = 0

masks, flows, styles = model.eval(image, batch_size=32, flow_threshold=flow_threshold, cellprob_threshold=cellprob_threshold, normalize={"tile_norm_blocksize": tile_norm_blocksize},)

fig = plt.figure(figsize=(12,5))
plot.show_segmentation(fig, image, masks, flows[0])
plt.tight_layout()
plt.show()
```
